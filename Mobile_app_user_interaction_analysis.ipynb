##Outliers

```python
#Outlier
Q3 = df["device_os_version"].quantile(0.75)
Q1 = df["device_os_version"].quantile(0.25)
IQR = Q3 - Q1
IQR
```

```python
low_limit = Q1 - 1.5*IQR
upp_limit = Q3 + 1.5*IQR
low_limit , upp_limit
```

```python
df = df[(df["device_os_version"] >=low_limit) & (df["device_os_version"]<=upp_limit)]
```

##Exploratory Data Analysis (EDA)

```python
#1. Which device OS has the highest number of user interactions?

# Count Plot :
plt.figure(figsize=(7,5))
sns.countplot(data=df, x='device_os')
plt.xlabel("Device OS")
plt.ylabel("Number of Interactions")
plt.title("User Interactions by Device OS")
plt.show()
```

```python
# Observation:
# Android has significantly more interactions than iOS.
# it indicating that Android users represent the larger share of activity in the dataset.
```

```python
# 2. What percentage of users are subscribed?
# Pie Chart:
plt.figure(figsize=(5,6))
df['is_subscribed'].value_counts().plot(kind='pie',autopct='%1.1f%%',labels=['Subscribed', 'Not Subscribed'])
plt.ylabel("")
plt.title("Subscribed vs Not Subscribed Users")
plt.show()
```

```python
# Observation:
# The subscription distribution is almost evenly split, with slightly more than half of the users subscribed.
# This indicates that there is still a substantial opportunity to convert non-subscribed users.
```

```python
# 3. Which event types are most frequently performed by users?
# Bar Chart :
plt.figure(figsize=(10,6))
event_counts = df['event_type'].value_counts()
sns.barplot(x=event_counts.values,y=event_counts.index)
plt.xlabel("Number of Interactions")
plt.ylabel("Event Type")
plt.title("User Interactions by Event Type")
plt.show()
```

```python
# Observation:
# Click is the most common interaction, followed by share and view.
# The event counts are relatively balanced.
# showing that users perform a wide range of interactions rather than relying on only one activity.
```

```python
# 4.Top 10 device models generate the most interactions?
# Horizontal Bar Chart :
plt.figure(figsize=(10,6))
top_devices = df['device_model'].value_counts().head(10)
sns.barplot(x=top_devices.values, y=top_devices.index)
plt.xlabel("Number of Interactions")
plt.ylabel("Device Model")
plt.title("Top 10 Device Models by User Interactions")
plt.show()
```

```python
# Observation:
# Xiaomi Mi 11 has the highest number of interactions among the device models.
# Android devices dominate the top interaction-generating models.
```

```python
# 5.Top 10 event targets that receive the most user interactions?
# Bar Chart :
plt.figure(figsize=(10,6))
top_targets = df['event_target'].value_counts().head(10)
sns.barplot(x=top_targets.values, y=top_targets.index)
plt.xlabel("Number of Interactions")
plt.ylabel("Event Target")
plt.title("Top Event Targets")
plt.show()
```

```python
# Observation:
# The checkout button receives the highest number of interactions.
# This suggests that users frequently reach or interact with the purchasing/conversion area of the application.
```

```python
# 6. What is the correlation between the main numerical variables?
# Heatmap :
plt.figure(figsize=(9,6))
numeric_cols = ['battery_level','memory_usage_mb','event_value','session_duration_sec','user_age']
corr = df[numeric_cols].corr()
sns.heatmap(corr,annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap of Numerical Variables")
plt.show()
```

```python
# Observation:
# The numerical variables have very weak correlations with each other.
# Therefore, user age, battery level, event value, and session duration do not appear to have strong linear relationships in this dataset.
```
